## Soil water retention curve
$
S(\psi)=\frac{\theta-\theta_{res}}{\left(1+\left|\alpha\psi\right|^n\right)^{m_{vg}}} + \theta_{res}, \quad m_{vg} = 1-\frac{1}{n_{vg}}
$

where:

- ($\theta$): liquid water content
- ($\theta_r$): theta_res
- ($\theta_s$): theta_sat
- ($n$): $vGn_n$
For unsaturated soil (($\psi<0$), matric head), SUMMA uses van Genuchten-Mualem style forms (equivalent to code in soil_utils.f90):

$[
\theta(\psi)=\theta_r+(\theta_s-\theta_r)\left(1+\left|\alpha\psi\right|^n\right)^{-m}
]$

$[
K(\psi)=K_{sat};
\frac{(\theta_s-\theta_r)}
{\left(1+\left|\alpha\psi\right|^n\right)^{m/2}}
]$

with $\alpha = vGn_{\alpha}$ magnitude units ($m^{-1}$)

How they affect behavior:

- $vGn_{\alpha}$ (air-entry scale)
  - Larger ($|\alpha|$): lower (\theta) at same suction, weaker capillary hold, faster drainage/percolation, earlier runoff response, less late-season storage.
  - Smaller ($|\alpha|$): stronger capillary retention, slower drainage, longer recession.
- $vGn_n$ (curve shape / pore-size spread)
  - Larger n: steeper retention curve near saturation, conductivity changes more abruptly, often flashier wet-season response and quicker drying once threshold crossed.
  - Smaller n: smoother retention and conductivity transition, more gradual release.

For your goal (reduce late-summer retained water), the usual direction is:

increase (|\alpha|) modestly
increase n modestly
co-tune with k_soil (often dominant) 

## Big Bucket aquifer:

$q_{bf} = q_{ref}\left(\frac{S_{aq}}{S_{scale}}\right)^{b}$

- ($q_{bf}$): aquifer baseflow flux (scalarAquiferBaseflow, m s(^{-1}))
- ($S_{aq}$): aquifer storage (scalarAquiferStorage, m)
- ($S_{scale}$): aquiferScaleFactor (or basin__aquiferScaleFactor), m
- ($b$): aquiferBaseflowExp (or basin__aquiferBaseflowExp), dimensionless
- ($q_{ref}$): reference baseflow coefficient
- local-param version: aquiferBaseflowRate (comment says “baseflow rate when aquifer storage = aquiferScaleFactor”)
  - this is the $q_{ref}$
- basin-param version often uses basin__aquiferHydCond as the coefficient term in basin mode

So at $S_{aq}=S_{scale}, q_{bf}=q_{ref}$
